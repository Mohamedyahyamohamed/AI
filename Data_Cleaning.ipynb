## what columns mean ?

علومات أساسية عن المريضة (Patient Basic Information)


Sl. No (Serial Number):

المعنى: الرقم التسلسلي. هو مجرد رقم لترتيب وتنظيم البيانات في الجدول (مثل 1, 2, 3, ...).

Patient File No. (Patient File Number):

المعنى: رقم ملف المريضة. هو رقم تعريفي فريد لكل مريضة في النظام الطبي أو المستشفى.

PCOS (Y/N) (Polycystic Ovary Syndrome):

المعنى: متلازمة تكيس المبايض (نعم/لا). هذا العمود يوضح ما إذا كانت المريضة قد تم تشخيصها بمتلازمة تكيس المبايض أم لا. (Y = Yes = نعم، N = No = لا).

Age (yrs):

المعنى: العمر (بالسنوات). عمر المريضة وقت تسجيل البيانات.

Weight (Kg):

المعنى: الوزن (بالكيلوجرام). وزن المريضة.

Height (Cm):

المعنى: الطول (بالسنتيمتر). طول المريضة.

BMI (Body Mass Index):

المعنى: مؤشر كتلة الجسم. هو مقياس لتقييم ما إذا كان وزن الشخص صحيًا بالنسبة لطوله. يتم حسابه بالمعادلة:

BMI= الطول (بالمتر) /الوزن (بالكيلوجرام)
​


ملاحظة: يجب تحويل الطول من السنتيمتر إلى المتر قبل الحساب (بقسمته على 100).

Blood Group:

المعنى: فصيلة الدم. يحدد فصيلة دم المريضة (مثل A+, O-, B+).

المؤشرات الحيوية ونمط الحياة (Vital Signs & Lifestyle)


Pulse rate(bpm) (Beats Per Minute):

المعنى: معدل النبض (نبضة في الدقيقة). عدد دقات قلب المريضة في الدقيقة الواحدة.

RR (breaths/min) (Respiratory Rate):

المعنى: معدل التنفس (نفس في الدقيقة). عدد الأنفاس التي تأخذها المريضة في الدقيقة.

Fast food (Y/N):

المعنى: تناول الوجبات السريعة (نعم/لا). يوضح ما إذا كانت المريضة تستهلك الوجبات السريعة بانتظام.

Reg.Exercise(Y/N) (Regular Exercise):

المعنى: ممارسة الرياضة بانتظام (نعم/لا). يوضح ما إذا كانت المريضة تمارس التمارين الرياضية بشكل منتظم.

BP _Systolic (mmHg) (Systolic Blood Pressure):

المعنى: ضغط الدم الانقباضي (مليمتر زئبقي). هو الرقم الأعلى في قراءة ضغط الدم، ويمثل الضغط في الشرايين عندما ينبض القلب.

BP _Diastolic (mmHg) (Diastolic Blood Pressure):

المعنى: ضغط الدم الانبساطي (مليمتر زئبقي). هو الرقم الأسفل في قراءة ضغط الدم، ويمثل الضغط في الشرايين بين نبضات القلب (عندما يكون القلب في حالة راحة).

بيانات الفحص بالموجات فوق الصوتية (Ultrasound Data)
هذه الأعمدة مهمة جدًا لتشخيص متلازمة تكيس المبايض.

Follicle No. (L) (Follicle Number - Left):

المعنى: عدد البصيلات (في المبيض الأيسر). عدد البصيلات (الحويصلات الصغيرة التي تحتوي على البويضات) التي تم رصدها في المبيض الأيسر عبر فحص الموجات فوق الصوتية.

Follicle No. (R) (Follicle Number - Right):

المعنى: عدد البصيلات (في المبيض الأيمن). عدد البصيلات التي تم رصدها في المبيض الأيمن.

Avg. F size (L) (mm) (Average Follicle Size - Left):

المعنى: متوسط حجم البصيلات (في المبيض الأيسر) بالمليمتر. متوسط حجم هذه البصيلات في المبيض الأيسر. في حالة تكيس المبايض، غالبًا ما تكون البصيلات صغيرة الحجم.

Avg. F size (R) (mm) (Average Follicle Size - Right):

المعنى: متوسط حجم البصيلات (في المبيض الأيمن) بالمليمتر. متوسط حجم البصيلات في المبيض الأيمن.

Endometrium (mm):

المعنى: سُمك بطانة الرحم (بالمليمتر). قياس سُمك النسيج المبطن للرحم. يتغير هذا السُّمك خلال الدورة الشهرية، وقياسه يساعد في تقييم صحة الدورة الهرمونية.

## Data Cleaning means and steps



تنظيف البيانات (Data cleaning) هو عملية تجهيز البيانات للتحليل / تعلم الآلة (ML) / التعلم العميق (DL) عن طريق إزالة أو تعديل البيانات غير الصحيحة، أو غير المكتملة، أو غير ذات الصلة، أو المكررة، أو المنسقة بشكل غير سليم.

image.png



الخطوات الأساسية في عملية تنظيف وتجهيز البيانات (Data Cleaning and Preparation). إليك شرح بسيط لكل خطوة:

Handling Missing Data (التعامل مع البيانات المفقودة) 🤔
هذه الخطوة تعني إيجاد حل للخانات الفارغة في جدول البيانات. عندما تكون بعض المعلومات ناقصة، يجب أن نقرر كيف سنتعامل معها. يمكننا إما حذف الصف بأكمله، أو ملء الفراغ بقيمة تقديرية مثل المتوسط الحسابي للعمود.

Outlier Detection and Handling (اكتشاف ومعالجة القيم الشاذة) 🎯
القيم الشاذة (Outliers) هي بيانات تختلف بشكل كبير جدًا عن بقية البيانات. على سبيل المثال، إذا كانت لديك قائمة بأعمار موظفين (25, 30, 28, 65, 32)، فإن العمر "65" يعتبر قيمة شاذة. هذه القيم يمكن أن تؤثر سلبًا على دقة التحليل، لذلك يجب اكتشافها وتحديد إذا كان يجب إزالتها أو تعديلها.

Data Scaling and Transformation (تغيير نطاق البيانات وتحويلها) ⚖️
تهدف هذه العملية إلى توحيد مقياس الأرقام في الأعمدة المختلفة. فمثلًا، إذا كان لديك عمود للراتب (بالآلاف) وعمود للعمر (بالعشرات)، فإن الأرقام الكبيرة في الراتب قد تطغى على تأثير العمر. عملية الـ Scaling تضع كل البيانات في نطاق موحد (مثل من 0 إلى 1)، مما يضمن أن كل متغير يساهم بشكل عادل في التحليل.

Encoding Categorical Variables (ترميز المتغيرات الفئوية) 🔡➡️🔢
الخوارزميات ونماذج تعلم الآلة تفهم الأرقام وليس النصوص. المتغيرات الفئوية هي بيانات نصية مثل (أحمر، أخضر، أزرق) أو (مصر، السعودية، الإمارات). عملية الترميز (Encoding) هي تحويل هذه الكلمات إلى أرقام (مثلاً: مصر=1، السعودية=2) حتى يتمكن الكمبيوتر من معالجتها.

Handling Duplicates (التعامل مع البيانات المكررة) 👯
هذه هي عملية البحث عن الصفوف المكررة بالكامل في البيانات وإزالتها. وجود بيانات مكررة يمكن أن يسبب انحيازًا في النتائج ويجعل بعض الأنماط تبدو أكثر أهمية مما هي عليه في الواقع.

Dealing with Inconsistent Data (معالجة البيانات غير المتسقة) ✅
البيانات غير المتسقة هي البيانات التي تحتوي على تناقضات أو أخطاء إملائية. على سبيل المثال، قد تجد نفس المدينة مكتوبة بأشكال مختلفة في نفس العمود مثل ("Cairo", "cairo", "Ciro"). هذه الخطوة تهدف إلى توحيد هذه البيانات وتصحيح الأخطاء لجعلها متسقة وصحيحة.

### remove outlier usinig  IQR

```python

Q1=pcos['BMI'].quantile(0.25)
Q3=pcos['BMI'].quantile(0.75)
IQR=Q3-Q1
upperIBM=Q3+1.5*IQR
lowerIBM=Q1-1.5*IQR
```

```python
pcos['BMI'] = np.where(pcos['BMI'] > upperIBM, upperIBM, np.where(pcos['BMI'] < lowerIBM, lowerIBM, pcos['BMI']))
```

```python
sns.boxplot(data=pcos,x='BMI')
plt.show()
```

```python
def cap_outliers(df, column_name):
    Q1 = df[column_name].quantile(0.25)
    Q3 = df[column_name].quantile(0.75)
    IQR = Q3 - Q1
    upper_bound = Q3 + 1.5 * IQR
    lower_bound = Q1 - 1.5 * IQR

    df[column_name] = np.where(df[column_name] > upper_bound, upper_bound,
                           np.where(df[column_name] < lower_bound, lower_bound, df[column_name]))
    return df


columns_to_clean = ['Follicle No. (L)', 'Follicle No. (R)', ' Age (yrs)', 'BMI']

for col in columns_to_clean:
    print(f"Processing outliers for column: {col}")
    pcos = cap_outliers(pcos, col)
print("Outlier processing complete.")

plt.figure(figsize=(12, 10))
for i, col in enumerate(columns_to_clean):
    plt.subplot(2, 2, i + 1)
    sns.boxplot(x=pcos[col], showmeans=True)
    plt.title(f'Distribution of {col} after Capping')
plt.tight_layout()
plt.show()
```
